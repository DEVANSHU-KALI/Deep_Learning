### Now lets explore about what is MLP (multi layer perceptron) and AF (activation function)
In the lesson 1, we got to know how does a single neuron acts as straight line (y=mx+b). But here is a flaw about this, no matter how many straight lines you add, the mathematical result will always be another straight line. 

If ai could only draw a straight line, it could never draw a circle around something like face or anything. Real world data is curved, messy and non-linear. 

To solve this, DL has two massive upgrades.
1. Hidden layers (Multi-layer perceptron): we stack neurons horizontally and vertically, to create a network.
2. Activation function (The line bender): we put mathematical filters between the layers which physically bend the lines, allowing the network to mold itself around any shape of data.

#### The king of activation function: ReLU (Rectified Linear Unit)
There are so many activation functions (Sigmoid, Tanh), but 99% of modern AI (including LLMs) uses ReLU

The math of ReLU is very simple:
*If the number is negative, make it zero(0), if positive, just leave it alone*

That’s it. That tiny rule—chopping off the negative numbers—creates a "bend" or a "hinge" in the math. When you stack thousands of these bent lines together, a neural network can approximate any shape in the universe.

#### The Code: Building a Neural Network in PyTorch
In Lesson 1, we used a single layer nn.Linear(1, 1).

Now, we are going to build an actual Deep Learning architecture using PyTorch's standard format: the nn.Module class.

Here is how you build a Multi-Layer Perceptron (MLP) with one hidden layer and a ReLU activation function:


```python
import torch
import torch.nn as nn

# STEP 1: Define the Architecture (The Blueprint)
class SimpleMLP(nn.Module):
    def __init__(self):
        super(SimpleMLP, self).__init__()
        
        # Layer 1 (Hidden Layer): Takes 1 input, passes it to 10 hidden neurons
        self.layer1 = nn.Linear(in_features=1, out_features=10)
        
        # The Activation Function: The "line bender"
        self.relu = nn.ReLU()
        
        # Layer 2 (Output Layer): Takes the 10 neurons, squashes them to 1 prediction
        self.layer2 = nn.Linear(in_features=10, out_features=1)

    def forward(self, x):
        # This defines exactly how the data flows through the network
        x = self.layer1(x)  # Draw 10 straight lines
        x = self.relu(x)    # Bend the lines at zero
        x = self.layer2(x)  # Combine the bent lines into a final prediction
        return x

# STEP 2: Instantiate the Network
model = SimpleMLP()

# Let's test it with a dummy input!
di = torch.tensor([[3.0]])
prediction = model(di)

print("Network Architecture:\n", model)
print("\nPrediction from random weights:", prediction.item())
```

Step-by-Step Breakdown:

1. class SimpleMLP(nn.Module): This is the gold standard for PyTorch. Every custom neural network you ever write (even the massive GPT model) will inherit from nn.Module.
2. __init__ (The Setup): This is where you declare the parts of your engine. We created a layer with 10 neurons, a ReLU function, and an output layer.
3. forward (The Execution): This is the sequence of operations. The data x flows into Layer 1, gets bent by ReLU, and gets squeezed into a single output by Layer 2.

If you didn't put self.relu(x) in the middle, Layer 1 and Layer 2 would mathematically collapse into a single straight line, making the 10 hidden neurons completely useless.
