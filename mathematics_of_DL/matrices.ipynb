matrix multiplication is a fundamental operation in linear algebra, where two matrices are combined to produce a new matrix. The process involves taking the dot product of rows from the first matrix with columns from the second matrix.

we need to have the basics of matrix multiplication, including the rules and conditions for multiplying matrices.

Calculate profit/loss from revenue and expenses

```python
import numpy as np

revenue = np.array([[180,200,220],[24,36,40],[12,18,20]])
expenses = np.array([[80,90,100],[10,16,20],[8,10,10]])
```

```python
profit = revenue - expenses
profit
```


Calculate total sales from units and price per unit using matrix multiplication

```python

price_per_unit = np.array([1000,400,1200])
units = np.array([[30,40,50],[5,10,15],[2,5,7]])
```

```python
price_per_unit*units
```

In above case numpy is using broadcasting so it expands price_per_unit array from 1 row, 3 columns to 3 row and 3 columns. Correct way to do matrix multiplication is to use dot product as shown below

```python
np.dot(price_per_unit,units)
```
