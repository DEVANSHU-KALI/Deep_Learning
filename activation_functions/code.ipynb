lets see how **sigmoid** function works

```python
import math

def sigmoid(x):
    """Compute the sigmoid function."""
    return 1 / (1 + math.exp(-x))
```

```python
sigmoid(1)
```

```python
sigmoid(-19)
```

```python
sigmoid(100)
```

no matter what number you give to sigmoid function, it squeezes that value between 0 and 1

---

now lets see how **Tanh** works

```python
def tanh(x):
    """Compute the hyperbolic tangent function."""
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))
```

```python
tanh(1)
```

```python
tanh(-19)
```

```python
tanh(100)
```

all the values are squeezed between -1 and 1

now lets see how **relu** works

```python
def relu(x):
    """Compute the ReLU function."""
    return max(0, x)
```

```python
relu(1)
```

```python
relu(-19)
```

```python
relu(100)
```

if a value be less than 0, its converted into 0 and whatever value be greater than 1, its stays the same. 

---

now lets see **leaky relu**

```python
def leaky_relu(x):
    return max(0.1 * x, x)
```

```python
leaky_relu(1)
```

```python
leaky_relu(-19)
```

```python
leaky_relu(-200)
```

```python
leaky_relu(100)
```

instead of making any value to zero if the value is less than zero, leaky relu just converts the value just less than zero instead of totally to zero.
